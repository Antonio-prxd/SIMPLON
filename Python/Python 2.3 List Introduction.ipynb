```python
for i, val in enumerate(names):
  print(val)
```

2. What does `i` correspond to?  
What does `val` correspond to?  
Why is it so?

Answer: 

i : l'indice (la position) de l'élément dans la liste. Par défaut, il commence à 0.

val : la valeur de l'élément à cette position.

La fonction enumerate() permet de récupérer à la fois la position et la valeur de chaque élément lors d'une boucle.

```python
for val, i in enumerate(names):
  print(val,i)
```

3. This time, `i` returns first names and `val` returns numbers. Why is it so?

Answer: C’est parce que enumerate() renvoie d’abord le numéro de l’indice, puis la valeur. En écrivant for val, i, on inverse l’ordre des variables : val reçoit donc les nombres et i reçoit les prénoms.

4. What does the built in function enumerate do?
In what cases can it be useful?

Answer: En Data Analysis, enumerate() peut être utile lorsque tu parcours des résultats ou des valeurs et que tu veux conserver leur position pour les identifier, les comparer ou afficher un résultat numéroté.

2. What can the `.split()` method do?  
What is the type of `christmas` before using `.split()`?

Answer: 

La méthode .split() permet de découper une chaîne de caractères en plusieurs éléments à l'aide d'un séparateur.

Avant d'utiliser la méthode .split(), la variable est de type str (chaîne de caractères). Après l'application de cette méthode, si le résultat est stocké dans une variable, cette nouvelle variable est de type list (liste). La variable d'origine reste inchangée.

```python
christmas = "Noël c'est bien, mais attention à ne pas manger trop de chocolat. On le sait, mais on le fait à chaque fois. C'est la vie"


christmas.split("à")
```

3. Why does the `.split()` method split the `string` into 3 elements?

Answer: 

La méthode .split("à") découpe la chaîne de caractères à chaque fois qu'elle rencontre le caractère "à".

Le caractère "à" apparaît deux fois dans la chaîne de caractères. La méthode .split("à") renvoie donc une liste contenant trois éléments.
